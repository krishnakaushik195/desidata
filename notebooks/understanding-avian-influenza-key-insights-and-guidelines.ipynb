Avian influenza, commonly known as bird flu, is a zoonotic viral disease that primarily affects poultry but can occasionally spill over to mammals, including humans and swine. Notable subtypes of influenza type A viruses include H1N1, H3N2, and H5N1. The typical incubation period for avian influenza in humans ranges from 3 to 5 days and can extend to 7 to 10 days. Management of avian influenza cases is similar to that of influenza caused by primary human influenza viruses. While vaccines against the A(H5N1) virus have been developed and approved in multiple countries, their availability remains limited, and decisions on usage depend on infection risk assessments.

## Mini baseline — can TF-IDF find the source paragraph?

Each question was written from its row's source paragraph. A simple TF-IDF
retriever has to rank the correct one first. Its hit rate is the number
any trained model on this dataset must beat.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
import numpy as np

sample = df.head(300)
questions = sample["question"].astype(str).to_numpy()
passages = sample["context"].astype(str).to_numpy()

vectorizer = TfidfVectorizer(stop_words="english")
passage_vectors = vectorizer.fit_transform(passages)
question_vectors = vectorizer.transform(questions)

best = np.empty(len(questions), dtype=int)
for i in range(len(questions)):
    best[i] = linear_kernel(question_vectors[i], passage_vectors).argmax()

hit_rate = (passages[best] == passages).mean()
print(f"TF-IDF retrieval lands on the correct paragraph for {hit_rate:.0%} of questions")
print("That is the number a trained model has to beat.")
```
