# Economic Survey 2025-26 — Question and Answer Dataset

Generated starter notebook from [DesiData](https://www.desidata.in) — India's open dataset catalogue. Run the cells top to bottom.

849 extractive question-and-answer pairs built from the Economic Survey 2025-26, published by
the Department of Economic Affairs, Ministry of Finance, Government of India.

The source analyses India's economic performance and policy measures against global trends,
covering domestic and global growth, fiscal and monetary policy, sectoral performance across
manufacturing, services and agriculture, trade and investment including exports, imports and FDI,
infrastructure, employment and skilling, innovation, sustainability and climate finance,
education reform, and human capital.

Every answer is a verbatim span of text the Survey prints. Each row carries the passage the
answer sits in, the character offset of the answer within that passage, the exact source quote,
the page number and the location inside the document, so any row can be checked against the
original PDF.

175 of the 849 pairs (20.6%) are explanatory questions — definitions, cause-and-effect
relationships, summaries, comparisons and lists — and 674 restate a figure or a table cell.
99.76% of rows pass the corpus quality gate with zero blocking findings.

Coverage note: the Survey is long enough that 63.5% of it was read closely (325 of 512 parts).
The dataset therefore reflects a sample spread evenly across the document rather than the whole
of it. 8,468 of 13,514 text blocks and 84 of 286 tables were used.

## Mini baseline — can TF-IDF find the source paragraph?

Each question was written from its row's source paragraph. A simple TF-IDF
retriever has to rank the correct one first. Its hit rate is the number
any trained model on this dataset must beat.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
import numpy as np

sample = df.head(300)
questions = sample["question"].astype(str).to_numpy()
passages = sample["context"].astype(str).to_numpy()

vectorizer = TfidfVectorizer(stop_words="english")
passage_vectors = vectorizer.fit_transform(passages)
question_vectors = vectorizer.transform(questions)

best = np.empty(len(questions), dtype=int)
for i in range(len(questions)):
    best[i] = linear_kernel(question_vectors[i], passage_vectors).argmax()

hit_rate = (passages[best] == passages).mean()
print(f"TF-IDF retrieval lands on the correct paragraph for {hit_rate:.0%} of questions")
print("That is the number a trained model has to beat.")
```
