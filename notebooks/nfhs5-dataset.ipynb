## Load the dataset

The DesiData API URL is public and permanent — no API key required.

```python
import pandas as pd

url = "https://www.desidata.in/api/datasets/nfhs5-dataset/download"
df = pd.read_csv(url)

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")
df.head()
```

## Clean up numeric types

Comma-formatted numbers often load as text. Anything that is
overwhelmingly numeric becomes a real number here.

```python
for col in df.columns:
    cleaned = df[col].astype(str).str.replace(",", "", regex=False)
    converted = pd.to_numeric(cleaned, errors="coerce")
    if converted.notna().sum() >= 0.8 * max(df[col].notna().sum(), 1):
        df[col] = converted

print("Numeric columns after cleanup:", df.select_dtypes("number").columns.tolist())
```

## Visualise

Numeric columns: `value`, `source_page`

```python
import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes("number").columns.tolist()
print("Numeric columns:", numeric_cols)

df["value"].plot(kind="hist", bins=25, figsize=(10, 4), title="Distribution of value")
plt.tight_layout()
plt.show()
```

## Compare across `indicator`

Group means are the fastest way to spot a pattern worth investigating.

```python
category_col = "indicator"

top = df.groupby(category_col)["value"].mean().sort_values(ascending=False).head(15)
top.plot(kind="bar", figsize=(11, 4), title="Average value by indicator")
plt.tight_layout()
plt.show()
```
