## Load the dataset

The DesiData API URL is public and permanent — no API key required.

```python
import pandas as pd

url = "https://www.desidata.in/api/datasets/economic-survey-2025-26-question-and-answer-dataset/download"
df = pd.read_csv(url)

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")
df.head()
```

## Visualise

A first look at whatever the data offers most naturally.

```python
import matplotlib.pyplot as plt

# Comma-formatted numbers often load as text — convert columns that are overwhelmingly numeric.
for col in df.columns:
    cleaned = df[col].astype(str).str.replace(",", "", regex=False)
    converted = pd.to_numeric(cleaned, errors="coerce")
    if converted.notna().sum() >= 0.8 * max(df[col].notna().sum(), 1):
        df[col] = converted

numeric_cols = df.select_dtypes("number").columns.tolist()
text_cols = df.select_dtypes("object").columns.tolist()

if numeric_cols:
    df[numeric_cols[0]].plot(kind="hist", bins=25, figsize=(10, 4), title=f"Distribution of {numeric_cols[0]}")
elif text_cols:
    df[text_cols[0]].value_counts().head(15).plot(kind="barh", figsize=(9, 5), title=f"Top values in {text_cols[0]}")
else:
    print("Nothing obvious to chart — the exploration above is your starting point.")
plt.tight_layout()
plt.show()
```
